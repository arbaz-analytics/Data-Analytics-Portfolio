## EMPLOYEE SALES PERFORMANCE ANALYTICS SYSTEM

```python
# Collect the data : 

employees = {
    101: {
        "name": "Rahul",
        "region": "North",
        "department": "Electronics",
        "sales": 85000,
        "target": 80000,
        "rating": 4.3
    },

    102: {
        "name": "Arbaz",
        "region": "West",
        "department": "Electronics",
        "sales": 125000,
        "target": 100000,
        "rating": 4.9
    },

    103: {
        "name": "John",
        "region": "South",
        "department": "Furniture",
        "sales": 65000,
        "target": 70000,
        "rating": 3.9
    },

    104: {
        "name": "Priya",
        "region": "East",
        "department": "Office Supplies",
        "sales": 98000,
        "target": 90000,
        "rating": 4.5
    }
}
# EXPLORATORY DATA ANALYSIS (EDA)

# Sprint 01 : Employee Details
print("EMPLOYEES DETAIL")
for emp_id , detail in employees.items():
    print("=" * 40)
    print(f"Employee ID :        {emp_id}")
    print(f"Employee Name :      {detail['name']}")
    print(f"Region      :        {detail['region']}")
    print(f"Department :         {detail['department']}")
    print(f"Total Sales :        {detail['sales']}")
    print(f"Target :             {detail['target']}")
    print(f"Employee Rating :    {detail['rating']}")
    print("=" * 40)

# Sprint 02 : Total Employee Sales

def total_sales(employees):
    total = 0
    for emp, detail in employees.items():
        total += detail['sales']
    return total
result = total_sales(employees)
print(f"\nTotal Sales : ${result}")

# Sprint 03 : Average Employee Sales

def average_sales(employees):
    average = total_sales(employees)/len(employees)
    return average
avg_result = average_sales(employees)
print(f"\nAverage Sales : ${avg_result}\n")

# Sprint 04 : Highest Employee Sales
 
def highest_sales(employees):
 
    highest = 0
    employee_id = ""
    employee_name = ""
    for employee , detail in employees.items():
        if detail['sales'] > highest:
            highest = detail['sales']
            employee_id = employee
            employee_name = detail['name']
   
    return (f"Name : {employee_name} - Highest Sales :  ${highest} ")
print(highest_sales(employees))

# Sprint 05 : Lowest Employee Sales

def lowest_sales(employees):
    lowest = float('inf')  # It make sure that the very first number in a databset will always be smaller than it.
    employee_id = ""
    employee_name = ""
    for employee , detail in employees.items():
        if detail['sales'] < lowest:
            lowest = detail['sales']
            employee_id = employee
            employee_name = detail['name']
    return (f"\nName : {employee_name}  - Lowest Sales :  ${lowest} ")
print(lowest_sales(employees))


# Sprint 06 : Employees Above Average

def above_average(employees):
    
    total_value = sum(detail['sales'] for detail in employees.values())
    average_value = total_value / len(employees)
    
    above_avg_list = []
    above_avg_list.append(f"\nEmployees Above Average :")
    above_avg_list.append(f"{"=" * 50}")
    for employee , detail in employees.items(): 
        if detail['sales'] > average_value:
            above_avg_list.append(f"Name : {detail['name']} - Sales : ${detail['sales']:.2f}")
    above_avg_list.append(f"{"="*50}")
    return '\n'.join(above_avg_list)

print(above_average(employees))


    
# Sprint 07 : Department wise sales analysis

def departments(employees):
    
    department_data = {}
    for employee , detail in employees.items():
        department = detail['department']
        if department not in department_data:
            department_data[department] = {"sales":0,"count":0}
        department_data[department]["sales"] += detail['sales']
        department_data[department]["count"] += 1
  
    dept_result = ["\nDepartment - Wise Sales Performance:",f"{"="*40}\n"] 
    for dept , info in department_data.items():
        dept_result.append(f"Department : {dept}")
        dept_result.append(f"Total Sales :   ${info['sales']:.2f}")
        average = info['sales'] / info['count']
        dept_result.append(f"Average Sales : ${average}\n")
        dept_result.append(f"{"="*40}\n")

    return '\n'.join(dept_result)

print(departments(employees))

# Sprint 08 : Top 3 Best Performers(Employees)

def top_employees(employees):
    top_performers = sorted(employees.items() , key = lambda sale : sale[1]['sales'] , reverse = True)
    
    top_list = []
    top_list.append(f"Top Performance Employees")
    top_list.append(f"{'='*30}")
    for emp_id , detail in top_performers[:3]:
        top_list.append(f"Name: {detail['name']} - Sales: {detail['sales']}")
    top_list.append(f"{'='*30}\n")
    
    return '\n'.join(top_list)
    
print(top_employees(employees))

# Sprint 09 -  Sales Performance Report Generator
def sales_report(employees):
    report = [
        "="*40,
        f"Employees Sales Performance Report",
        "="*40,
        f"Total Employees : {len(employees)}",
        f"Total Sales : {total_sales(employees)}",
        f"Average Sales : {average_sales(employees)}",
        f"Employee Above Average : {len(above_average(employees))}",
        f"\nHighest Performer :\n{highest_sales(employees)}",
        f"\nLowest Performer :{lowest_sales(employees)}",
        f"{departments(employees)}" 
             ]

    return '\n'.join(report)
print(sales_report(employees))

# Sprint 10 : Employee Searching System

def employee_search(employees):
    
    print('='*30)
    print('Search for Employees :')
    print('='*30)
    
    print("1.Search by ID")
    print("2.Search by Name")
    choice = int(input("Choose one 1 or 2 :"))
    
    if choice == 1:
        search_id = int(input("Enter Employee ID : "))
        for emp_id , detail in employees.items():
            if search_id == emp_id:
                return display_employee(emp_id,detail) 
                
    elif choice == 2:
        search_name = input("Enter Employee Name : ").lower().strip()
        for emp_id , detail in employees.items():
            if search_name == detail['name'].lower().strip():
                return display_employee(emp_id,detail)
            
    return "Employee not found!"

    
def display_employee(emp_id , detail):
    lines = [
           '='*30 ,
           f"Employee ID : {emp_id}",
           f"Name: {detail['name']}",
           f"Sales: {detail['sales']}",
           f"Department: {detail['department']}",
           f"Region: {detail['region']}",
           f"Rating : {detail['rating']}",
           '='*30
         ]
    
    return "\n".join(lines)
    
print(employee_search(employees))

# Sprint 11 - Employees Rating System

def employee_rating(employees):
    
    print('Employees Sales Performance Rating')
    for emp_id , detail in employees.items():
        sales = detail['sales']
        
        if sales >= 100000:
            detail['rating'] = "Excellent"
        elif sales >= 50000:
            detail['rating'] = "Good"
           
        elif sales >= 30000:
            detail['rating'] = "Average"
            
        else:
            detail['rating'] = "Needs Improvement"
            
        print(display_employee(emp_id, detail))
        
employee_rating(employees)

# Sprint 12 - Employee Sales Analytics System 

def employee_analytics(employees):
    
    while True:
        print('='*35)
        print("Employee Sales Analytics System")
        print('='*35)
        menu_list = ["1.Total Sales", "2.Highest Performer", "3.Lowest Performer",
                     "4.Above Average Employees", "5.Department Analytics", 
                     "6.Top 3 Employees", "7.Search Employee", "8.Employee Rating", "9.Complete Sales Report", "0.Exit"]
        for menu in menu_list:
            print(menu)
    

        choice = int(input("Enter Your Choice: "))
        if choice == 1:
            print(f"Total Sales: {total_sales(employees)}")
            input("\nPress Enter to return to the menu...")
        elif choice == 2:
            print(f"{highest_sales(employees)}")
            input("\nPress Enter to return to the menu...")
        elif choice == 3:
            print(f"{lowest_sales(employees)}")
            input("\nPress Enter to return to the menu...")
        elif choice == 4:
            print(above_average(employees))
            input("\nPress Enter to return to the menu...")
        elif choice == 5:
            print(departments(employees))
            input("\nPress Enter to return to the menu...")
        elif choice == 6:
            print(top_employees(employees))
            input("\nPress Enter to return to the menu...")
        elif choice == 7:
            print(employee_search(employees))
            input("\nPress Enter to return to the menu...")
        elif choice == 8:
            employee_rating(employees)
            input("\nPress Enter to return to the menu...")
        elif choice == 9:
            print(sales_report(employees))
            input("\nPress Enter to return to the menu...")
        elif choice == 0:
            print("Thank you for using Employee Sales Analytics System")
            break
            
employee_analytics(employees)
            
            
```
